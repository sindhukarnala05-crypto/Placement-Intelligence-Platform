```python
import pandas as pd
import random
from faker import Faker
fake = Faker('en_IN')
branches = [
    "CSE",
    "CSM",
    "AIML",
    "DS",
    "ECE",
    "EEE",
    "IT"
]
students = []
for i in range(1, 201):
    students.append({
        "StudentID": i,
        "FullName": fake.name(),
        "Gender": random.choice(["Male", "Female"]),
        "Branch": random.choice(branches),
        "CGPA": round(random.uniform(6.5, 9.9), 2),
        "GraduationYear": 2027,
        "Email": fake.email(),
        "Phone": fake.msisdn()[:10]
    })
students_df = pd.DataFrame(students)
students_df.head()
```

```python
students_df.to_csv("students.csv", index=False)
```

```python
student_skills = []

skill_levels = ["Beginner", "Intermediate", "Advanced"]

student_skill_id = 1

for student in range(1,201):

    skills = random.sample(range(1,16), random.randint(3,6))

    for skill in skills:

        student_skills.append({
            "StudentSkillID": student_skill_id,
            "StudentID": student,
            "SkillID": skill,
            "SkillLevel": random.choice(skill_levels)
        })

        student_skill_id += 1

student_skills_df = pd.DataFrame(student_skills)

student_skills_df.head()
```

```python
student_skills_df.to_csv("studentskills.csv",index=False)
```
